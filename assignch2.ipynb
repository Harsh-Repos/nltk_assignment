```python
#6
text = "Hello, how aree you , 123?"

one = nltk.re_show(r'[a-zA-Z]+', text)

two = nltk.re_show(r'[A-Z][a-z]*', text)

three = nltk.re_show(r'p[aeiou]{,2}t', text)

four = nltk.re_show(r'\d+(\.\d+)?', text)

five = nltk.re_show(r'([^aeiou][aeiou][^aeiou])*', text)

six = nltk.re_show(r'\w+|[^\w\s]+', text)


print(one)
print(two)
print(three)
print(four)
print(five)
print(six)
```

```python
#21
from nltk.corpus import words

def unknown(url):
    link = request.urlopen(url).read().decode("utf-8")

    link_words = set(re.findall(r'\b[a-z]+\b', link))

    eng_words = set(words.words())

    unkown_words = link_words-eng_words
    return sorted(unkown_words)

url = "https://www.nltk.org/book/ch03.html"
print(unknown(url))
```

```python
#24
text = "I like to eat chocolate. Some people smile."

text = text.lower()

text = re.sub(r'ate', '8', text)
text = re.sub(r'e', '3', text)
text = re.sub(r'i', '1', text)
text = re.sub(r'o', '0', text)
text = re.sub(r'l', '|', text)
text = re.sub(r's', '5', text)
text = re.sub(r'\.', '5w33t!', text)

print(text)
```

```python
#26
text = """alma ember iskola ablak utca barack magyar asztal kutya"""

words = text.lower().split()

vowels = "aeiou"
vowel_seq = []

for word in words:
    seq = "".join(re.findall("["+vowels+"]", word))
    vowel_seq.append(seq)

print(vowel_seq)
```

```python
#30
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

text = "The students are studying"
tokens = word_tokenize(text)

stemmer = PorterStemmer()

for word in tokens:
    stemmed = stemmer.stem(word)
    print(word, "-", stemmed)
```
