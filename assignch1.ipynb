```python
#5
from nltk.corpus import brown
nltk.download("brown")
```

```python
humor = brown.words(categories="humor")
romance = brown.words(categories="romance")


humor_total = len(humor)
humor_unique = len(set(humor))
humor_diversity = humor_unique / humor_total


romance_total = len(romance)
romance_unique = len(set(romance))
romance_diversity = romance_unique / romance_total

print(humor_diversity)
print(romance_diversity)
```
